```python
#General Data Visualizations
# Distribution of Life Expectancy
plt.figure(figsize=(8, 6))
sns.histplot(df['Life expectancy'], kde=True)  # Histogram with KDE
plt.title('Distribution of Life Expectancy')
plt.xlabel('Life Expectancy')
plt.show()
```

```python

# Calculate the average life expectancy for each year
yearly_avg = df.groupby('Year')['Life expectancy'].mean()

# Find the year with the lowest and highest average life expectancy
lowest_year = yearly_avg.idxmin()
highest_year = yearly_avg.idxmax()

print(f"Year with the lowest average life expectancy: {lowest_year}")
print(f"Year with the highest average life expectancy: {highest_year}")
```

```python

#1. How do education levels and welth impact life expectancy?

#Life Expectancy vs. Schooling
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Schooling', y='Life expectancy', data=df)
plt.title('Life Expectancy vs. Schooling')
plt.xlabel('Schooling')
plt.ylabel('Life Expectancy')
plt.show()

# Life Expectancy vs. GDP
plt.figure(figsize=(8, 6))
sns.scatterplot(x='GDP', y='Life expectancy', data=df)
plt.title('Life Expectancy vs. GDP')
plt.xlabel('GDP')
plt.ylabel('Life Expectancy')
plt.xlim(0, 60000)  # Set the x-axis limit
plt.show()
```

```python
#2 What are the countries with high (above the avg) Alcohol consumption and life expectance rate less than 60 years old?
import plotly.express as px
df_clean = df.dropna(subset=['Life expectancy', 'Alcohol', 'Country'])

# Calculate thresholds
avg_alcohol = df_clean['Alcohol'].mean()
life_cutoff = 60

# Filter target countries
high_risk = df_clean[
    (df_clean['Life expectancy'] < life_cutoff) & 
    (df_clean['Alcohol'] > avg_alcohol)
].groupby('Country')[['Alcohol', 'Life expectancy']].mean().reset_index()

# Create map plot
fig = px.choropleth(
    high_risk,
    locations="Country",
    locationmode="country names",
    color="Alcohol",
    hover_name="Country",
    hover_data={
        "Alcohol": ":.1fL",
        "Life expectancy": ":.1f years",
        "Country": False
    },
    color_continuous_scale=px.colors.sequential.Reds,
    title="Countries Drinking Above Average With Life Expectancy Below 60 Years"
)


fig.show()

# Print detailed list
print("Countries with High Alcohol Consumption and Low Life Expectancy:")
print("---------------------------------------------------------------")
print(f"{'Country':<20} | {'Alcohol/L':<10} | {'Life Expectancy':<15}")
print("-"*45)
for _, row in high_risk.sort_values('Alcohol', ascending=False).iterrows():
    print(f"{row['Country']:<20} | {row['Alcohol']:<10.1f} | {row['Life expectancy']:<15.1f}")
```

```python
#3.How death under-five hits poor countries and rich countries during years? 

# 1. Clean Data
# Fill missing values with column averages
df['under-five deaths '] = df['under-five deaths '].fillna(df['under-five deaths '].mean())
df['GDP'] = df['GDP'].fillna(df['GDP'].mean())

# 2. Categorize Countries
# Create wealth categories using GDP median
df['Wealth Category'] = ['Rich' if gdp >= df['GDP'].median() else 'Poor' for gdp in df['GDP']]

# 3. Prepare Visualization
plt.figure(figsize=(10, 6))
sns.set_style("whitegrid")

# Create the plot
sns.lineplot(
    data=df,
    x='Year',
    y='under-five deaths ',
    hue='Wealth Category',
    estimator='mean',
    errorbar=None,
    marker='o'
)

# Add labels and title
plt.title("Child Mortality Comparison: Rich vs Poor Countries", pad=20)
plt.xlabel("Year", labelpad=10)
plt.ylabel("Average Under-5 Deaths", labelpad=10)
plt.legend(title='Country Wealth')

plt.show()

# 4. Show Correlation
correlation = df['GDP'].corr(df['under-five deaths '])
print(f"\nStrong inverse relationship: GDP vs Child Deaths\nCorrelation Coefficient: {correlation:.2f}")
```

```python
#4 how adis affect some selected countried and how dose is affect there life expectancy ?
import plotly.express as px


# Selected countries for comparison
countries = ['South Africa', 'Zimbabwe', 'Eswatini', 'Nigeria', 'Kenya', 'Uganda', 'Russia','Germany', 'France', 'United Kingdom', 'Italy', 'Spain']

# Calculate average HIV rates
hiv_data = (
    df[df['Country'].isin(countries)]
    .groupby('Country', as_index=False)['HIV/AIDS']
    .mean()
)

# Create interactive map
px.choropleth(
    hiv_data,
    locations='Country',             # Country names
    locationmode='country names',    # Geographic mapping
    color='HIV/AIDS',                # Color scale based on HIV rates
    title='HIV/AIDS Rates Comparison: Selected Countries',
    color_continuous_scale='Plasma'  # Purple-yellow color scheme
).show()

#now lets show the avg life of the counties in the map to know if the AIDS afect there life expactnce 
selected_countries_df = df[df['Country'].isin(countries)]
avg_life_expectancy = selected_countries_df.groupby('Country')['Life expectancy'].mean()
plt.figure(figsize=(6, 4))
avg_life_expectancy.plot(kind='bar')
plt.title('Average Life Expectancy for Selected Countries')
plt.xlabel('countries')
plt.ylabel('Average Life Expectancy')
plt.show()
```

```python
#5 how life expactnce  rate affected by the good health care countries:'Japan', 'Canada', 'Australia' and bad ones :'Nigeria', 'Chad', 'Sierra Leone'?
# Define countries to compare
good_health = ['Japan', 'Canada', 'Australia']  # Countries with strong healthcare
poor_health = ['Nigeria', 'Chad', 'Sierra Leone']  # Countries with weak healthcare
all_countries = good_health + poor_health

# Filter and clean data
clean_data = df[df['Country'].isin(all_countries)]  # Keep only our target countries
clean_data = clean_data.dropna(subset=['Life expectancy'])  # Remove missing values

# Calculate average life expectancy per country
avg_life_by_country = clean_data.groupby('Country')['Life expectancy'].mean().reset_index()

# Set up plot
plt.figure(figsize=(10, 6))  # Adjust figure size for better readability of bar chart

# Plot bar chart
bars = plt.bar(avg_life_by_country['Country'], avg_life_by_country['Life expectancy'])

# Color-code the bars
for bar, country in zip(bars, avg_life_by_country['Country']):
    color = 'green' if country in good_health else 'red' #green for good health countries 
     #red for bad health counntries 
    #set the colors for the bars
    bar.set_color(color)

# Add plot labels and styling
plt.title("Green = Good Healthcare, Red = Poor Healthcare")
plt.xlabel("Country")
plt.ylabel("Average Life Expectancy (years)")
# Show the plot
plt.show()
# Print final averages for comparison
print("Average Life Expectancies:")
print(avg_life_by_country.set_index('Country'))
```

```python
#6.How does the income composition of resources, in conjunction with population density influence life expectancy

import seaborn as sns
df['Life expectancy'] = df['Life expectancy'].fillna(df['Life expectancy'].mean())
df['Income composition of resources'] = df['Income composition of resources'].fillna(
    df['Income composition of resources'].mean()
)
df['Population'] = df['Population'].fillna(df['Population'].mean())

#calculating in addition to grouping values over time 
grouped_time = df.groupby('Year')[['Life expectancy', 'Income composition of resources', 'Population']].mean()

#Visualizing in the form of line plotting over time 
plt.figure(figsize=(12, 6))
plt.plot(grouped_time.index, grouped_time['Life expectancy'], label='Life Expectancy')
plt.title('Life Expectancy Over Time')
plt.xlabel('Year')
plt.ylabel('Life Expectancy')
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(12, 6))
plt.plot(grouped_time.index, grouped_time['Income composition of resources'], label='Income Composition')
plt.title('Income Composition of Resources Over Time')
plt.xlabel('Year')
plt.ylabel('Income Composition')
plt.legend()
plt.grid(True)
plt.show()
    
plt.figure(figsize=(12, 6))
plt.plot(grouped_time.index, grouped_time['Population'], label='Population')
plt.title('Population Over Time')
plt.xlabel('Year')
plt.ylabel('Population')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#7.what is the biggest postive change in lif expec rate in one country ?
#8.what is the biggest nigtive change in lif expec rate in one country ?
# Group by country and year, then calculate the mean life expectancy for each year
grouped_df = df.groupby(['Country', 'Year'])['Life expectancy'].mean().reset_index()

# Get the first and last year for each country
first_year = grouped_df.groupby('Country')['Year'].transform('min')
last_year = grouped_df.groupby('Country')['Year'].transform('max')

# Get the life expectancy in the first and last year
first_life_expectancy = grouped_df[grouped_df['Year'] == first_year].set_index('Country')['Life expectancy']
last_life_expectancy = grouped_df[grouped_df['Year'] == last_year].set_index('Country')['Life expectancy']

# Calculate the change in life expectancy
change_df = pd.DataFrame({
    'Life Expectancy Change': last_life_expectancy - first_life_expectancy
})
```

```python

# 1.Find the country with the biggest increase
c1 = change_df.idxmax()[0]
biggestP_df = grouped_df[grouped_df['Country'] == c1]



# Plotting the results
# Plot life expectancy over time for the country with the biggest increase
plt.figure(figsize=(10, 6))
sns.lineplot(x='Year', y='Life expectancy', data=biggestP_df, marker='o')
plt.title(f"Life Expectancy Trend for {c1} (Biggest Increase)")
plt.xlabel('Year')
plt.ylabel('Life Expectancy')
plt.show()


# 4. Print the names of the countries with the biggest changes
print(f"Country with the biggest positive change: {c1}")

```

```python
# 2. Find the country with the biggest decrease in life expectancy
# Find the country with the biggest decrease
c2 = change_df.idxmin()[0]
biggestN_df = grouped_df[grouped_df['Country'] == c2]


# Plot life expectancy over time for the country with the biggest decrease
plt.figure(figsize=(10, 6))
sns.lineplot(x='Year', y='Life expectancy', data=biggestN_df, marker='o')
plt.title(f"Life Expectancy Trend for {c2} (Biggest Decrease)")
plt.xlabel('Year')
plt.ylabel('Life Expectancy')
plt.show()

print(f"Country with the biggest negative change: {c2}")
```

```python
#9.What are the top 10 Countries with most “Life Expectancy” increase between (2000-2015)?

# Calculate life expectancy changes
change = df.groupby('Country')['Life expectancy'].agg(['first', 'last'])
change['Increase'] = change['last'] - change['first']
top_10 = change.nlargest(10, 'Increase').reset_index()

# Create visualization
plt.figure(figsize=(10, 5))
plt.barh(top_10['Country'], top_10['Increase'], color='royalblue')
plt.gca().invert_yaxis()  # Show biggest at top
plt.title('Top 10 Life Expectancy Improvements (2000-2015)')
plt.xlabel('Years Gained')
plt.tight_layout()
plt.show()

# Print results
print("Biggest Improvements:")
print(top_10[['Country', 'Increase']])
```

```python
#10.what is the effect of thinness  1-19 years and thinness 5-9 years' at the Developing countries at 2000 and 2015?

# Filter for Developing countries in 2000 and 2015
thin_df = df[(df['Status'] == 'Developing') & (df['Year'].isin([2000, 2015]))]

# Fill missing values with column means
thin_df['thinness  1-19 years'].fillna(thin_df['thinness  1-19 years'].mean(), inplace=True)
thin_df['thinness 5-9 years'].fillna(thin_df['thinness 5-9 years'].mean(), inplace=True)

# Calculate the average thinness for each year
avg_1_19 = thin_df.groupby('Year')['thinness  1-19 years'].mean()
avg_5_9 = thin_df.groupby('Year')['thinness 5-9 years'].mean()

# Plot
years = [2000, 2015]
x = range(len(years))

plt.figure(figsize=(8, 5))
plt.bar([i - 0.2 for i in x], avg_1_19, width=0.4, label='Thinness 1–19 years', color='blue')
plt.bar([i + 0.2 for i in x], avg_5_9, width=0.4, label='Thinness 5–9 years', color='red')

plt.xticks(x, years)
plt.xlabel('Year')
plt.ylabel('Average % Thinness')
plt.title('Thinness Levels in Developing Countries (2000 vs 2015)')
plt.legend()

plt.show()

```
